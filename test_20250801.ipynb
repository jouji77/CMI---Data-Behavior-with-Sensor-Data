```python
test_df = pd.read_csv(fr"{datafol}\{test_file}")
test_demographics_df = pd.read_csv(fr"{datafol}\{test_demographics_file}")
train_df = pd.read_csv(fr"{datafol}\{train_file}")
train_demographics_df = pd.read_csv(fr"{datafol}\{train_demographics_file}")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score

skf = StratifiedKFold(n_splits=5)
for fold, (train_idx, val_idx) in enumerate(skf.split(train_df, train_df["behavior"])):
    train_data = train_df.iloc[train_idx]
    val_data = train_df.iloc[val_idx]
    # 特徴量とターゲットの選択
    feature_cols = [col for col in train_data.columns if col.startswith(('acc_', 'rot_', 'thm_', 'tof_'))]
    X_train = train_data[feature_cols]
    y_train = train_data['behavior']

    # 欠損値処理（例: -1を0に置換）
    X_train = X_train.replace(-1, 0)

    # モデル構築・学習
    clf = RandomForestClassifier()
    clf.fit(X_train, y_train)

    # 新規データで予測
    X_new = val_data[feature_cols].replace(-1, 0)
    pred_behavior = clf.predict(X_new)

    # 予測結果と正解ラベルの比較
    y_test = val_data['behavior']
    accuracy = accuracy_score(y_test, pred_behavior)
    print(f"Accuracy: {accuracy:.3f}")
```
