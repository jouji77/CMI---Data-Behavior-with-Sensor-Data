```python
# 身体データをtrain_dfに結合
train_df["adult_child"] = 0
train_df["age"] = 0
train_df["sex"] = 0
train_df["handedness"] = 0
train_df["height_cm"] = 0
train_df["shoulder_to_wrist_cm"] = 0
train_df["elbow_to_wrist_cm"] = 0

for idx, row in train_demographics_df.iterrows():
    train_df.loc[train_df["subject"] == row["subject"], "adult_child"] = row["adult_child"]
    train_df.loc[train_df["subject"] == row["subject"], "age"] = row["age"]
    train_df.loc[train_df["subject"] == row["subject"], "sex"] = row["sex"]
    train_df.loc[train_df["subject"] == row["subject"], "handedness"] = row["handedness"]
    train_df.loc[train_df["subject"] == row["subject"], "height_cm"] = row["height_cm"]
    train_df.loc[train_df["subject"] == row["subject"], "shoulder_to_wrist_cm"] = row["shoulder_to_wrist_cm"]
    train_df.loc[train_df["subject"] == row["subject"], "elbow_to_wrist_cm"] = row["elbow_to_wrist_cm"]
    
```

```python
train_df
```

```python
# 加速度のベクトル合成
train_df["acc_magnitude"] = np.sqrt(
    train_df["acc_x"]**2 + train_df["acc_y"]**2 + train_df["acc_z"]**2
)

group = train_df.groupby(['sequence_id', 'subject'])["acc_magnitude"]
train_df["acc_magnitude_mean"] = group.transform("mean")
train_df["acc_magnitude_std"] = group.transform("std")
train_df["acc_magnitude_max"] = group.transform("max")
train_df["acc_magnitude_min"] = group.transform("min")
train_df["acc_magnitude_range"] = train_df["acc_magnitude_max"] - train_df["acc_magnitude_min"]

train_df["jerk"] = (
    train_df
    .groupby(["sequence_id", "subject"])["acc_magnitude"]
    .diff()
)
```

```python
# 0〜8列目（代表値）: groupbyキーは除外
meta_cols = [c for c in train_df.columns[:9] if c not in ['sequence_id', 'subject']]

# センサー列（特徴量抽出対象）
sensor_cols = train_df.columns[9:21]

# 0〜8列目は各グループの最初の行を取得
meta_df = (
    train_df
    .groupby(['sequence_id', 'subject'], as_index=False)[meta_cols]
    .first()
)

# センサー列は統計量を計算（キー列を保持）
agg_funcs = {col: ['mean', 'std', 'max', 'min'] for col in sensor_cols}
sensor_df = (
    train_df
    .groupby(['sequence_id', 'subject'], as_index=False)
    .agg(agg_funcs)
)

# 列名をフラット化（キー列はそのまま）
sensor_df.columns = [
    col if isinstance(col, str) else '_'.join(col).strip()
    for col in sensor_df.columns
]

# range列を追加
for col in sensor_cols:
    sensor_df[f'{col}_range'] = sensor_df[f'{col}_max'] - sensor_df[f'{col}_min']

sensor_df = sensor_df.rename(columns={"sequence_id_": "sequence_id", "subject_": "subject"}) 

# 両方をマージ（キー列が確実に存在する）
train_df_cal = pd.merge(meta_df, sensor_df, on=['sequence_id', 'subject'])
```

```python
train_df_cal.head()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score

skf = StratifiedKFold(n_splits=3)
for fold, (train_idx, val_idx) in enumerate(skf.split(train_df_cal, train_df_cal["gesture"])):
    train_data = train_df_cal.iloc[train_idx]
    val_data = train_df_cal.iloc[val_idx]
    # 特徴量とターゲットの選択
    feature_cols = [col for col in train_data.columns if col.startswith(('acc_', 'rot_', 'thm_', 'tof_'))]
    X_train = train_data[feature_cols]
    X_train = X_train.astype(np.float32)
    y_train = train_data['gesture']

    # 欠損値処理（例: -1を0に置換）
    X_train = X_train.replace(-1, 0)

    # モデル構築・学習
    clf = RandomForestClassifier()
    clf.fit(X_train, y_train)

    # 新規データで予測
    X_new = val_data[feature_cols].replace(-1, 0)
    pred_gesture = clf.predict(X_new)

    # 予測結果と正解ラベルの比較
    y_test = val_data['gesture']
    accuracy = accuracy_score(y_test, pred_gesture)
    print(f"Accuracy: {accuracy:.3f}")
```
